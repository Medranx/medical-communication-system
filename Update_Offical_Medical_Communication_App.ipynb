```python
def signup_user(patient_name, patient_username, patient_password,patient_email,patient_phone):
  """
This function registers a new user into the system.

Arguments:
  patient_name(string): The full name of the patient
  patient_username(string): The account username
  patient_password(string): The account password
  patient_email(string): The patient's email
  patient_phone(string): The patient's phone number

Returns:
  string: A message confirming the account creation or a error message.
"""
# Prevents one patient from overwriting another's account and of course using the same username
  if patient_username in patients:
        return "Sign-Up Failed: That username is already taken. Please choose another."
#To check that the phone is a string
  if type(patient_phone) != str:
    return "Error: Phone must be entered as a string (ex. '123456789')"
#Checking the phone only contains numbers and not letters
  numbers="0123456789"

  for char in patient_phone:
    if char not in numbers:
      return "Sign-up Failed: Phone must only contain numbers."

  #Checks the email to have the @ and incldue the '.'

  if "@" not in patient_email:
    return "Sign-Up Failed: Invalid email format (missing @)"

  if "." not in patient_email:
    return " Sign-Up Failed: Invalid email format."
  #Checks the password length, must be more than 12 characters

  if len(patient_password) <= 12:
    return "Sign-up Failed: Password must be longer than 12 characters"
  #Checks that there is special characters, has_special assumes there are none so it is false
  #until it goes through the function making it true.
  special_char="!@#$%^&*()_+"
  has_special = False

  for char in patient_password:
    if char in special_char:
      has_special = True

  if has_special == False:
    return "Sign-up Failed: Password must contain at least one special character."
  #Saves this new user to their own dictionary
  store_patient_data(patient_name, patient_username, patient_password,
                       patient_phone, patient_email)
  return f"Success: Account has been created. Welcome, {patient_name}!"
```

```python
def login_user(patient_username,patient_password):
  """
  This function verifies the user's identity by comparing the already stored username and password in the
  patients dictionary.

  Arguments:
      patient_username(string): The account username
      patient_password(string): The account password

  Return:
    string: A message confirming that the user has logged in or an error message.
  """
  global current_session # Allows the function to change the login state

  #Defensive Coding: checks if the user left the information empty
  if patient_username == "" or patient_password == "":
    return "Login Failed: Pleas enter both username and password."

  #Verifies the information the user entered and if it is stored in the patients dictionary comparing the
  #entered data to the stored data using '=='

  if patient_username not in patients:
        return "Login Failed: Incorrect username or password."

  if patient_password == patients[patient_username]["password"]:
        current_session = patient_username
        return f"Success: Login successful. Welcome back, {patient_username}!"
        # If anything is wrong in the username or the password, the access will be denied
  else:
        return "Login Failed: Incorrect username or password."
```

```python
def password_management(patient_username,old_password, new_password):
  """
  This function allows the user to change their current password by verifying their old password and checking their
  new password follows the requirements.

  Arguments:
    patient_username(string): The account username
    old_password(string): The current password the user wants to change
    new_password(string): The new passsword the user wants to set
    # database_password(string): The password stored in the database - Removed as it's a global variable

  Return:
    string: A message confirming the password has been updated, or an error message explaining why it failed
  """
  #Accesses the patients dictionary to get the old password
  global patients

  #Verifies the username is in the dictionary
  if patient_username not in patients:
        return "Password Update Failed: Username not found."
  #Verifies the old password is correct
  if old_password != patients[patient_username]["password"]:
        return "Password Update Failed: The old password you entered is incorrect."

  #Prevents using the same preivously used password
  if new_password == old_password:
        return "Password Update Failed: New password cannot be the same as the old password."

  #Checks the new password follows all requirements
  if len(new_password) <= 12:
    return "Password Update Failed: New password must be longer than 12 characters."
  #Checks that there is special characters, has_special assumes there are none so it is false
  #until it goes through the function making it true.
  special_char="!@#$%^&*()_+"
  has_special = False

  for char in new_password:
    if char in special_char:
      has_special = True

  if has_special == False:
    return "Sign-up Failed: Password must contain at least one special character."
  #Saves the new password to the dictionary
  patients[patient_username]["password"] = new_password
  return f"Success: Password has been updated for {patient_username}!"
```

```python
#Patient App

def patient_app_portal(choice, patient_name="", patient_username="", patient_password="", patient_email="", patient_phone="",date="",time="",department="",old_password="", new_password="", appointment_id="", confirm=False):
    """
    This function acts as the main interface for the patient, routing their requests
    to the specific registration, medical record, or appointment functions.

    Arguments:
        choice (string): The user's selection ('register', 'check results', or 'request appt')
        patient_name (string): The full name of the patient
        patient_username (string): The account username
        patient_password (string): The account password
        patient_email (string): The patient's email
        patient_phone (string): The patient's phone number
        date(string): The requested date for the appointment(MM/DD/YYYY)
        time(string): The requested time for the appointmnet(HH:MM AM/PM)
        department(string): The specific department the appointment is for
        old_password (string): Current password for the username.
        new_password (string): The new password wanting to be.
        appointment_id (string): The unique ID of an existing appointment
        confirm (boolean): Must be True to finalize an appointment cancellation.


    Returns:
        string: The result message from the specific department function called.
    """
    #Choice 1 registers the user calling the sign_up function
    if choice == "register":
        # This calls signup_user function with all the validation
        return signup_user(patient_name, patient_username, patient_password, patient_email, patient_phone)
    #Choice 2 Logins the patient
    elif choice == "login":
        # Calls your login function
        return login_user(patient_username, patient_password)
    #Choice 3 Calls Change Pass
    elif choice == "change password":
        # Calls to the password management function
        return password_management(patient_username, old_password, new_password)
     #Choice 4 Appointments
    elif choice == "schedule":
        # This calls the schedule appt function to make an appt
        return schedule_appointment(patient_username, date, time, department)
     #Choice 5 Updates the current Appointment
    elif choice == "update appt":
        # This calls the update appt function and updates it
        return update_appointment(patient_username, appointment_id, "", "", "", date, time, department)
    #Choice 6 Cancels the current Appointment
    elif choice == "cancel appt":
        # This calls the cancel appt fucntion
        return cancel_appointment(patient_username, appointment_id,confirm)
    #Choice 7 Checks patient's medical results
    elif choice == "check results":
      #Runs the view my records function to see the patients records
        return view_my_records(patient_username,"", "", "", "", "")
    #Choice 8 Checks the patient's prescriptions
    elif choice == "check prescription":
      #Runs the check my prescription function to see current prescriptions
        return check_my_prescriptions(patient_username,"", "", "")
    #Choice 9 Checks notification alerts
    elif choice == "check alerts":
      #Runs the check notifications functions to see current notifications
        return check_notifications(patient_username,"", "")
    else:
        return "Invalid selection. Please choose designated menu keywords."
```

```python
def schedule_appointment(patient_username,date,time,department):
  """
  This function will log a medical appointment for a patient

  Arguments:
    patient_username(string): The username of the patient
    date(string): The requested date for the appointment(MM/DD/YYYY)
    time(string): The requested time for the appointmnet(HH:MM AM/PM)
    department(string): The specific department the appointment is for

  Returns:
    string: Gives a message confirming that the appointment has been scheduled or an error message if data is missing
    """
  global appointment_counter

  #This will ensure that none of the information is left blank
  if patient_username =="" or date =="" or time=="" or department =="":
    return "Scheduling Failed: All fields are needed to schedule an appointment."

  #Checks the date is in correct format
  if len(date) != 10:
    return "Scheduling Failed: Please enter date as MM/DD/YYYY (ex. 08/09/2004)."

  #Checks the user specifies AM or PM so the system has no issues.
  if "AM" not in time and "PM" not in time:
    return " Scheduling Failed: Please specify AM or PM for your appointment time."

  #Generates a unique appointment ID
  appointment_counter += 1
  appointment_id = "APT-" + str(appointment_counter)
  #Saves it on the the patient's dictionary
  store_appointment_log(patient_username, date, time, department,
                          "Scheduled", appointment_id)
  #If information passes all the parameters then it will be saved to the dictionary
  return (f"Success: Appointment {appointment_id} has been scheduled for "
            f"{patient_username}! Save this ID to cancel or change your appointment.")

```

```python
def cancel_appointment(patient_username,appointment_id,confirm_orcancel):
  """
  This function cancels an existing appointment after the user confirms

  Arguments:
     patient_username(string): The username of the patient
     appointment_id(string): The specific ID of the appointment that will be canceled
     confirm_cancel(bool): A True/False that will practically confirm the cancelation

  Return:
    string: A message confirming the cancellation or an error message

  """

  #If the user doesn't confirm the confirm_orcancel this will prevent it from changing anything.
  if confirm_orcancel == False:
    return " Cancellation Failed: No changes were made to your current appointment."

  #Makes sure the user puts the appointment_id and the patient_username
  if patient_username =="" or appointment_id=="":
    return "Cancellation Failed: Missing Patient Username or Appointment ID."
  #Saves it onto the dictionary
  store_appointment_log(patient_username, "N/A", "N/A", "N/A",
                          "Cancelled", appointment_id)
  return "Success: Appointment has been cancelled."
```

```python
def update_appointment(patient_username,appointment_id,date,time,department,new_date,new_time,new_dept):
  """
  This function updates the current apppointment the patient has.

  Arguments:
    patient_username(string): The username of the patient.
    appointment_id(string):The specific ID of the appointment that will be canceled
    date(string): The requested date for the appointment(MM/DD/YYYY)
    time(string): The requested time for the appointmnet(HH:MM AM/PM)
    department(string): The specific department the appointment is for
    new_date(string): The new date of the appointment(MM/DD/YYYY)
    new_time(string): The new time of the appointment(HH:MM AM/PM)
    new_dept(string)L The new department for the new appointment

  Returns:
    string: A message confirming the updated details of the new appointment or an error message.
  """
    #To know what the patient wants to keep the appointment will only update what is filled, if something is left blank
    #The previous information will stay the same
  final_date = date if new_date == "" else new_date
  final_time = time if new_time == "" else new_time
  final_dept = department if new_dept == "" else new_dept


  #Checks all the new information for the new appointment meets the requirements.

  #Checks the date is in correct format
  if len(final_date) != 10:
    return "Scheduling Failed: Please enter date as MM/DD/YYYY (ex. 08/09/2004)."

  #Checks the user specifies AM or PM so the system has no issues.
  if "AM" not in new_time and "PM" not in new_time:
    return "Scheduling Failed: Please specify AM or PM for your appointment time."
  #Saves it all to the dictionary and if it works returns success
  store_appointment_log(patient_username, final_date, final_time, final_dept,
                          "Updated", appointment_id)
  return (f"Success: Appointment {appointment_id} is now set for "
            f"{final_dept} on {final_date} at {final_time}!")
```

```python
def main():
    """
    Main controller for the Medical Communication App.
    This will control all the functions and call them in the correct order needded
    and it will manage the whole interaction witht he user making it easier for them

    Arguments:
        None

    Returns:
        None
    """

    # Menu options - each keyword is connected to its corresponding function from the patient app to run them
    actions = {
        "register": handle_register,
        "login": handle_login,
        "change password": handle_change_password,
        "schedule": handle_schedule,
        "update appt": handle_update_appt,
        "cancel appt": handle_cancel_appt,
        "check results": handle_check_results,
        "check prescription": handle_check_prescription,
        "check alerts": handle_check_alerts,}
    running = True
    while running:
        display_patient_menu()

        try:
            choice = input("\n  Enter your choice: ").strip().lower()
        except (KeyboardInterrupt, EOFError):
            print("\n\n  Session interrupted. Goodbye!")
            break

        if choice == "0" or choice == "exit":

            print("  Thank you for using the Patient Portal. Stay healthy!")

            running = False

        elif choice in actions:
            try:
                actions[choice]()
            except Exception as error:
                print(f"\n  An unexpected error occurred: {error}")
                print("  Please try again.")

        else:
            print("\n  !! Invalid option. Please type one of the menu keywords.")

        if running:
            input("\n  Press Enter to return to the main menu...")



```
